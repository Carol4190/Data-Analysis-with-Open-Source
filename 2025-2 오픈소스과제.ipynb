```python
import requests
import time

# 1. 기본 설정
# 서울 열린데이터 광장에서 발급받은 본인의 API 키를 입력하세요.
API_KEY = "554b6f61716d73343531426"

# 데이터 조회 기간 설정
YEARS = range(2015, 2025)  # 2015년부터 2024년까지
MONTHS = range(1, 13)     # 1월부터 12월까지

print("--- API 월별 호출 상태 확인 시작 ---")

# 2. 반복 실행 및 URL 생성
for year in YEARS:
    for month in MONTHS:
        # 월은 항상 두 자리 숫자로 만들어줍니다 (예: 1 -> "01", 9 -> "09")
        month_str = f"{month:02d}"

        # 요청할 URL 동적 생성 (1개 구 데이터만 받아와서 확인 속도를 높임)
        url = f"http://openapi.seoul.go.kr:8088/{API_KEY}/json/energyUseDataSummaryInfo/1/1/{year}/{month_str}"

        # 3. API 호출 및 결과 확인 (핵심 로직)
        try:
            # API 서버에 GET 요청 (10초 이상 응답 없으면 타임아웃)
            response = requests.get(url, timeout=10)

            # --- 결과 판별 ---
            # 가. 서버와 통신이 성공했을 경우 (HTTP 상태 코드가 200)
            if response.status_code == 200:
                # API 서버가 보내준 JSON 데이터 확인
                data = response.json()
                # API 자체 결과 코드 추출
                api_result = data.get('energyUseDataSummaryInfo', {}).get('RESULT', {})
                api_code = api_result.get('CODE', '코드 없음')

                print(f"✅ {year}년 {month_str}월: 통신 성공 (HTTP {response.status_code}) / API 결과: {api_code}")

            # 나. 서버와 통신은 됐지만, 페이지가 없거나(404) 서버에 문제가 있을 경우
            else:
                print(f"❌ {year}년 {month_str}월: 통신 실패 (HTTP {response.status_code})")

        # 다. 네트워크 연결 문제 등 요청 자체가 실패했을 경우
        except requests.exceptions.RequestException as e:
            print(f"❌ {year}년 {month_str}월: 요청 중 에러 발생 - {e}")

        # 서버에 부담을 주지 않도록 각 요청 사이에 0.1초 대기
        time.sleep(0.1)

print("\n--- 모든 확인 작업 완료 ---")
```
