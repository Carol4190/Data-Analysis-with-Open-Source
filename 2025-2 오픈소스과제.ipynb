```python
import requests
import pandas as pd

# 1. API를 호출하여 데이터를 가져옵니다.
api_key = "554b6f61716d733435314261494b62"
url = f'http://openapi.seoul.go.kr:8088/{api_key}/json/energyUseDataSummaryInfo/1/7/2015/01'
# 참고: 1/5 -> 1/7로 변경하여 2015년 1월의 모든 유형(7개) 데이터를 가져오도록 했습니다.

response = requests.get(url)

# 데이터 프레임을 담을 변수 초기화
df = None

# 2. API 호출이 성공했는지 확인합니다. (HTTP 상태 코드 200)
if response.status_code == 200:
    print("✅ API 호출 성공!")

    # 3. 받아온 JSON 형식의 응답을 파이썬 딕셔너리로 변환합니다.
    data = response.json()

    # 4. 실제 데이터가 있는 부분('row')을 추출합니다.
    # JSON 구조: {'energyUseDataSummaryInfo': {'row': [...]}}
    try:
        rows = data['energyUseDataSummaryInfo']['row']

        # 5. 추출한 데이터를 pandas DataFrame으로 변환합니다.
        df = pd.DataFrame(rows)
        print("✅ JSON 데이터를 DataFrame으로 성공적으로 변환했습니다.\n")

    except KeyError:
        print("❌ 데이터 변환 실패: JSON 응답에서 'row' 키를 찾을 수 없습니다.")
        print("받은 데이터:", data)

else:
    print(f"❌ API 호출 실패: {response.status_code}")


# 6. DataFrame이 성공적으로 생성되었으면 기본 정보를 출력합니다.
if df is not None:
    print("--- [1] 데이터 샘플 (상위 5개) ---")
    print(df.head())
    print("\n" + "="*50 + "\n")

    print("--- [2] 데이터프레임의 기본 정보 (구조, 데이터 타입 등) ---")
    # .info()는 각 컬럼의 이름, 데이터 개수, 데이터 타입을 보여줍니다.
    df.info()
    print("\n" + "="*50 + "\n")

    print("--- [3] 숫자형 데이터의 기술 통계 요약 ---")
    # .describe()는 숫자 컬럼들의 개수, 평균, 표준편차, 최소/최대값 등을 보여줍니다.
    # 일부 컬럼은 숫자처럼 보이지만 문자열일 수 있으므로, 숫자형으로 변환 후 요약합니다.
    numeric_df = df.apply(pd.to_numeric, errors='coerce') # 숫자로 변환, 안되는 값은 NaT/NaN 처리
    print(numeric_df.describe())
```
